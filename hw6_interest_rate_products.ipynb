### Theory: EUR/USD FX Forward Collateralised in GBP

**Setup and notation.** Let $X(t)$ be the spot EUR/USD rate (USD per 1 EUR), $N_€$ the EUR notional, $K$ the agreed forward rate and $T_s$ the settlement date. The long party receives $N_€\,X(T_s)$ USD and pays $N_€\,K$ USD at $T_s$, i.e. a single USD cashflow

$$\Pi(T_s)=N_€\,\big(X(T_s)-K\big).$$

The trade is collateralised in GBP, a third currency. Write $r^{\$},r^{€},r^{£}$ for the risk-free (OIS) short rates, $r^{£}_c$ for the GBP collateral rate (SONIA) and $\lambda^{£}=r^{£}-r^{£}_c$ for the GBP funding spread. We choose USD as the reporting (pricing) currency, so $p=\$$, $\text{col}=£$ and $p\neq\text{col}$: the foreign-collateral formula applies.

For a currency $d$ and a collateral currency $c$ we use the collateralised discount factor

$$P^{d,c}(t,T)=\mathbb{E}^{Q^{d}}_t\!\left[e^{-\int_t^{T} r^{d}(s)\,ds}\;e^{\int_t^{T}\left(r^{c}(s)-r^{c}_c(s)\right)ds}\right],$$

the price at $t$ of 1 unit of currency $d$ paid at $T$ when the trade is collateralised in currency $c$. In particular $P^{d,d}$ is the domestic OIS discount factor.

---

#### 1. Valuation formula

Under the USD risk-neutral measure $Q^{\$}$, a USD payoff collateralised in GBP is valued by discounting at the USD risk-free rate and multiplying by the GBP collateral-funding factor (Piterbarg / Fujii et al.):

$$V^{\$}(t)=N_€\,\mathbb{E}^{Q^{\$}}_t\!\left[e^{-\int_t^{T_s} r^{\$}(s)\,ds}\;e^{\int_t^{T_s}\left(r^{£}(s)-r^{£}_c(s)\right)ds}\;\big(X(T_s)-K\big)\right].$$

Split the payoff into the $K$ term and the $X(T_s)$ term.

*The $K$ term* is a known USD amount, so by definition of the collateralised discount factor

$$\mathbb{E}^{Q^{\$}}_t\!\left[e^{-\int r^{\$}}e^{\int(r^{£}-r^{£}_c)}\,K\right]=K\,P^{\$,£}(t,T_s).$$

*The $X(T_s)$ term* is converted into EUR by changing measure from $Q^{\$}$ to $Q^{€}$ with the Radon–Nikodym derivative

$$\left.\frac{dQ^{€}}{dQ^{\$}}\right|_{t\to T_s}=\frac{X(T_s)\,B^{€}(t,T_s)}{X(t)\,B^{\$}(t,T_s)},\qquad B^{d}(t,T)=e^{\int_t^{T} r^{d}(s)\,ds},$$

which gives

$$\mathbb{E}^{Q^{\$}}_t\!\left[e^{-\int r^{\$}}e^{\int(r^{£}-r^{£}_c)}\,X(T_s)\right]
= X(t)\,\mathbb{E}^{Q^{€}}_t\!\left[e^{-\int r^{€}}e^{\int(r^{£}-r^{£}_c)}\right]
= X(t)\,P^{€,£}(t,T_s).$$

Therefore

$$\boxed{\,V^{\$}(t)=N_€\left[X(t)\,P^{€,£}(t,T_s)-K\,P^{\$,£}(t,T_s)\right]\,}$$

Interpretation: the EUR leg is worth $N_€$ EUR discounted on the GBP-collateralised EUR curve and converted at spot; the USD leg is $N_€K$ USD discounted on the GBP-collateralised USD curve. Dividing by $X(t)$ gives the same value expressed in EUR.

---

#### 2. Fair forward rate

The fair forward $K^{*}$ makes the contract worth zero at inception:

$$V^{\$}(t)=0\;\Longleftrightarrow\;\boxed{\,K^{*}=FX^{£}_{\$/€}(t,T_s)=X(t)\,\frac{P^{€,£}(t,T_s)}{P^{\$,£}(t,T_s)}\,}$$

This is the collateral-dependent forward $FX_{d/f}=X\,P^{f,col}/P^{d,col}$ with $d=\$$, $f=€$, $\text{col}=£$. The valuation formula can then be written as

$$V^{\$}(t)=N_€\,P^{\$,£}(t,T_s)\,\big(K^{*}-K\big),$$

i.e. the notional times the difference between the current fair forward and the contract forward, discounted on the GBP-collateralised USD curve.

---

#### 3. Are the five curves sufficient?

Map the given curves to the notation:

| Given curve | Meaning | Notation |
|---|---|---|
| dfSONIA | GBP OIS discounting | $P^{£,£}$ |
| dfEuroSTR | EUR OIS discounting | $P^{€,€}$ |
| dfSOFR | USD OIS discounting | $P^{\$,\$}$ |
| df[GBP,USD] | GBP cashflows collateralised in USD | $P^{£,\$}$ |
| df[EUR,USD] | EUR cashflows collateralised in USD | $P^{€,\$}$ |

The formulas require $P^{€,£}$ and $P^{\$,£}$, and **neither is given directly**: there is no liquid EUR/GBP cross-currency market to build $P^{€,£}$ from, and no GBP-collateralised USD curve. They must be obtained through the USD-centred currency chain.

**Assumption.** The forward FX rate does not depend on the collateral currency (equivalently, the cross-currency funding spreads are deterministic, or uncorrelated with FX, so the convexity correction $\epsilon=0$). Under this assumption the EUR/GBP forward computed with USD collateral equals the one computed with GBP collateral:

$$X_{£/€}\,\frac{P^{€,\$}}{P^{£,\$}}=X_{£/€}\,\frac{P^{€,£}}{P^{£,£}}
\quad\Longrightarrow\quad
P^{€,£}(t,T)=P^{€,\$}(t,T)\,\frac{P^{£,£}(t,T)}{P^{£,\$}(t,T)},$$

and likewise for the USD/GBP forward:

$$P^{\$,£}(t,T)=P^{\$,\$}(t,T)\,\frac{P^{£,£}(t,T)}{P^{£,\$}(t,T)}.$$

Substituting into the results of parts 1–2:

$$K^{*}=X(t)\,\frac{P^{€,\$}(t,T_s)}{P^{\$,\$}(t,T_s)}=X(t)\,\frac{\text{df[EUR,USD]}(T_s)}{\text{dfSOFR}(T_s)},$$

$$V^{\$}(t)=N_€\,\frac{\text{dfSONIA}(T_s)}{\text{df[GBP,USD]}(T_s)}\Big[X(t)\,\text{df[EUR,USD]}(T_s)-K\,\text{dfSOFR}(T_s)\Big].$$

**Conclusion.**

1. **Yes, the five curves are sufficient** under the currency-chain assumption: $P^{€,£}$ and $P^{\$,£}$ are both rebuilt from USD-collateralised curves and the GBP OIS curve.
2. Only four of the five curves are actually used. dfEuroSTR ($P^{€,€}$) does not enter the price of this trade — it would be needed only for EUR-collateralised trades or when calibrating the EUR/USD basis itself.
3. Sanity check: $K^{*}$ coincides with the USD-collateralised forward $X\,P^{€,\$}/P^{\$,\$}$. This is exactly the statement that the forward is independent of the collateral currency; the collateral currency affects only the discounting factor $P^{\$,£}=\text{dfSOFR}\cdot\text{dfSONIA}/\text{df[GBP,USD]}$ applied to the mark-to-market $K^{*}-K$.
4. If the assumption is dropped — i.e. the GBP funding spread $\lambda^{£}$ is stochastic and correlated with $X$ — the expectation in part 1 contains a covariance term $\mathrm{Cov}\!\left(e^{\int\lambda^{£}},\,X(T_s)\right)$ that no set of discount curves can reproduce; a joint model of rates, spreads and FX would be required, and the five curves would no longer be sufficient.

## 1. Input data

Rates are annual decimals: `0.04 = 4%`, `0.001 = 10 bp`. These are synthetic teaching inputs.

on top of SOFR, as in HW5. Edit the inputs here and rerun the following cells.

```python
VALUE_DATE = "2026-01-02"
EFFECTIVE_DATE = "2026-04-02"
FRA_END_DATE = "2026-07-02"
SWAP_END_DATE = "2027-04-02"
FUNDING_INDEX = "SOFR-1B-FLAT"
NOTIONAL = 1_000_000
FIXED_RATE = 0.04
BASIS_SPREAD = 0.001
BASIS_CONVENTION = "USD-FF-3M-OVER-USD-SOFR-OIS-3M"

sofr_quotes = pd.DataFrame(
    {"values": [0.03, 0.04, 0.05]},
    index=pd.Index(["2026-04-02", "2027-01-04", "2028-01-03"], name="Node date"),
)
ff_quotes = pd.DataFrame(
    {"values": [0.028, 0.037]},
    index=pd.Index(["2026-07-02", "2028-01-03"], name="Node date"),
)
funding_quotes = pd.DataFrame(
    {"values": [0.0, 0.001]},
    index=pd.Index(["2026-07-02", "2028-01-03"], name="Node date"),
)

display(sofr_quotes.rename(columns={"values": "SOFR-1B IFR"}))
display(ff_quotes.rename(columns={"values": "FF-1B IFR"}))
display(funding_quotes.rename(columns={"values": "Funding IFR spread"}))
```

## 2. Data and build methods

Create the IFR datasets and build methods, with SOFR first, Fed Funds second and funding third.
The model construction copies these IFR inputs into state variables, as in HW5.

```python
sofr_data = qfCreateData1D("INSTANTANEOUS FORWARD RATE", "SOFR-1B-IFR", sofr_quotes)
ff_data = qfCreateData1D("INSTANTANEOUS FORWARD RATE", "FF-1B-IFR", ff_quotes)
funding_data = qfCreateData1D(
    "INSTANTANEOUS FORWARD RATE", "SOFR-1B-FUNDING-IFR", funding_quotes,
)
data_collection = qfCreateDataCollection([sofr_data, ff_data, funding_data])
display(data_collection.display())
```

```python
sofr_method = qfCreateBuildMethod("YIELD_CURVE_INDEX", {
    "TARGET": "SOFR-1B", "INSTANTANEOUS FORWARD RATE": "SOFR-1B-IFR",
})
ff_method = qfCreateBuildMethod("YIELD_CURVE_INDEX", {
    "TARGET": "FF-1B", "INSTANTANEOUS FORWARD RATE": "FF-1B-IFR",
})
funding_method = qfCreateBuildMethod("YIELD_CURVE_FUNDING", {
    "TARGET": FUNDING_INDEX, "REFERENCE INDEX": "SOFR-1B",
    "INSTANTANEOUS FORWARD RATE": "SOFR-1B-FUNDING-IFR",
})
build_methods = qfCreateModelBuildMethodCollection([sofr_method, ff_method, funding_method])
```

## 3. Build the model

`FundingIndexParameter` selects the curve used to discount both legs on their payment dates.

```python
model = qfCreateModel(VALUE_DATE, "YIELD_CURVE", data_collection, build_methods)
funding_parameter = qfCreateValuationParameters(
    "FUNDING INDEX PARAMETER", {"FUNDING INDEX": FUNDING_INDEX},
)
parameters = qfCreateValuationParametersCollection([funding_parameter])

print("Component sizes:", model.num_sub_components)
print("Component order:", model.component_indices)
print("SOFR DF:", model.discount_factor("SOFR-1B", Date(FRA_END_DATE)))
print("Funding DF:", model.discount_factor(FUNDING_INDEX, Date(FRA_END_DATE)))
```

## 4. Interest-rate streams

## 5. FRA

The teaching FRA fixes the whole-period simple rate at $T_s$ and settles at $T_e$.
Before fixing, use the SOFR projection curve to calculate
$F=(P_p(t,T_s)/P_p(t,T_e)-1)/\tau$.
For receive-fixed, $PV=N\tau(K-F)D_f(t,T_e)$. `pay` reverses the sign.
The projection curve identifies a teaching term rate; its fixing is stored separately from daily SOFR fixings.

The FRA is supplied in full. Read `ProductFRA.__init__`, `ProductFactory.create_fra`,
and `ValuationEngineProductFra.calculate_value` to see how product construction,
convention-driven creation, and leg valuation fit together. Use these implementations
as the starting point for the two swap products.


```python
fra = qfCreateProductFRA(
    EFFECTIVE_DATE, FRA_END_DATE, "SOFR-1B", FIXED_RATE, "receive", NOTIONAL,
    accrual_basis="ACT/360",
)
display(qfDisplayProduct(fra))
display(qfCreateCashflowsReport(model, fra, parameters).display())
```

```python
df_start = model.discount_factor("SOFR-1B", fra.effective_date)
df_end = model.discount_factor("SOFR-1B", fra.termination_date)
forward_rate = (df_start / df_end - 1.0) / fra.accrued
expected_fra_pv = (
    NOTIONAL * fra.accrued * (FIXED_RATE - forward_rate)
    * model.discount_factor(FUNDING_INDEX, fra.payment_date)
)
fra_engine = ValuationEngineProductFra(model, parameters, fra, ValuationRequest.PV)
fra_engine.calculate_value()
print("Projected term rate:", forward_rate)
print("FRA PV / formula:", fra_engine.value, expected_fra_pv)
np.testing.assert_allclose(fra_engine.value, expected_fra_pv, atol=1e-8)
```

Once $T_s$ is reached, supply the observed term fixing and discount the known payment.
The following fixing is synthetic.


```python
fixing_model = qfCreateModel(EFFECTIVE_DATE, "YIELD_CURVE", data_collection, build_methods)
observed_term_rate = 0.045
qfInsertTermRateFixing("SOFR-1B", EFFECTIVE_DATE, FRA_END_DATE, "ACT/360", observed_term_rate)
fixed_fra_pv = qfCreateValueReport(fixing_model, fra, parameters, "PV")[0][1]
known_amount = NOTIONAL * fra.accrued * (FIXED_RATE - observed_term_rate)
print("Known amount at Te:", known_amount)
print("PV on Ts:", fixed_fra_pv)
np.testing.assert_allclose(
    fixed_fra_pv, known_amount * fixing_model.discount_factor(FUNDING_INDEX, fra.payment_date),
    atol=1e-8,
)
```
